```python
import pandas as pd
import numpy as np
orders    = pd.read_csv(r"file_path")
items     = pd.read_csv(r"file_path")
customers = pd.read_csv(r"file_path")
products  = pd.read_csv(r"file_path")
reviews   = pd.read_csv(r"file_path")
payments  = pd.read_csv(r"file_path")
```

```python
orders
```

```python
lis = ["order_purchase_timestamp" , "order_approved_at" , "order_delivered_carrier_date" , "order_delivered_customer_date" , "order_estimated_delivery_date"]

for item in lis : 
    orders[item] = pd.to_datetime(orders[item])
```

```python
orders["order_status"].value_counts()
```

```python
orders2 = orders[orders["order_status"] == "delivered"]
```

```python
orders2.isnull().sum()
```

```python
orders2 = orders2.dropna(subset = ["order_delivered_customer_date"])
```

```python
orders2.isnull().sum()
```

```python
orders2.shape
```

```python
reviews.head()
```

```python
rev_count = reviews.groupby(["order_id"]).agg(no_of_review = ("review_id" , "count"),
                                              median_review_score = ("review_score" , "median"))
```

```python
rev_count.head()
```

```python
rev_count.shape
```

```python
items.head()
```

```python
items_new = items.groupby("order_id").agg(Count_of_product = ("product_id" , "count"),
                                       Total_price_of_order = ("price" , "sum"),
                                       Shipping_price = ("freight_value" , "sum"))
                                        
```

```python
payments.head()
```

```python
payments_new = payments.groupby("order_id").agg(Total_payament_method = ("payment_type" , "nunique"),
                                Total_payment_value = ("payment_value" , "sum"))
```

```python
customers.shape
```

```python
orders2.shape
```

```python
orders2.head()
```

```python
main_df = pd.merge(left=customers , right=orders2 , on="customer_id" ,how= "right")
```

```python
main_df.shape
```

```python
main_df.head()
```

```python
rev_count.reset_index(inplace=True)
items_new.reset_index(inplace=True)
payments_new.reset_index(inplace=True)
```

```python
main_df = pd.merge(left=main_df , right=rev_count , on="order_id" ,how= "left").copy()
main_df = pd.merge(left=main_df , right=items_new , on="order_id" ,how= "left").copy()
main_df = pd.merge(left=main_df , right=payments_new , on="order_id" ,how= "left").copy()
```

```python
main_df.shape
```

```python
main_df.isnull().sum()
```

```python
main_df.head()
```

```python
main_df= main_df.dropna(subset=["no_of_review" , "median_review_score" , "Total_payment_value"]).copy()
```
